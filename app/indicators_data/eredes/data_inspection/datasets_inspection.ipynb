## Observaciones sobre el dataset '8-total-upac-mensal':


Se comprueba que el dataset presenta 18 columnas, todas ellas con las unidades correctas según lo establecido en el catálogo.
La columna cpes hace referencia al Código do Ponto de Entrega, un identificador único, digital e invariable de 20 dígitos que la distribuidora asigna a cada punto de suministro de energía. Si bien es poco probable que se utilice directamente en este proyecto, era la única variable cuya interpretación faltaba por determinar.

En cuanto al tipo de dato de cada columna, se ha adelantado la transformación de data de texto a datetime para poder determinar correctamente el rango temporal del dataset. Respecto al resto de columnas, los códigos se mantienen como variables identificadoras no numéricas, puesto que no tiene sentido realizar operaciones matemáticas o de agregación sobre ellos. Por otro lado, se podrían transformar a categóricas las columnas tipo_de_tecnologia, nivel_de_tensao y escalao_de_potencia_instalada. En el caso de nivel_de_tensao y escalao_de_potencia_instalada, además, podría establecerse un orden interno si resulta útil para el análisis.

En cuanto a las observaciones que se pueden extraer de la función describe(), se aprecia una dispersión muy elevada en potencia_instalada_upac_kw: mientras que la mediana es de 35 kW y el percentil 75 se sitúa en 98,12 kW, el máximo alcanza los 19.600 kW, lo que indica una distribución fuertemente sesgada hacia valores altos. Antes de considerar estos valores como errores, habría que estudiar si pueden deberse a grandes instalaciones energéticas o a agregaciones de múltiples instalaciones dentro de una misma observación.

Para determinar si estos valores extremos son reales, conviene revisar las filas concretas y comprobar su coherencia con el número de instalaciones, la potencia por instalación, el resto de variables del registro y la evolución temporal de esa misma zona. Un valor extremo puede ser perfectamente válido si responde a una instalación o conjunto de instalaciones de gran tamaño; en cambio, podría considerarse sospechoso si resulta incoherente con el resto de información disponible.

Algo similar sucede con numero_de_instalacoes, que presenta una mediana de 2 instalaciones, un percentil 75 de 9 y un máximo de 2.389 instalaciones. Este valor máximo puede ser perfectamente real si determinadas zonas concentran una gran cantidad de instalaciones UPAC. La elevada dispersión de la potencia instalada también se refleja en relacao_potencia_por_cpe, que presenta igualmente una distribución muy sesgada.

También habría que investigar el hecho de que el mínimo de potencia_instalada_upac_kw sea 0 mientras que el mínimo de numero_de_instalacoes sea 1, ya que esto implicaría la existencia de registros con al menos una instalación pero una potencia total registrada de 0 kW.

En cuanto a los 1.511 valores nulos de distrito, concelho, cpes y las relaciones calculadas a partir de CPE, es muy probable que estén relacionados. Conviene comprobar si los valores faltantes corresponden exactamente a las mismas filas. Dado que los códigos territoriales no presentan valores nulos, en primer lugar se podría intentar recuperar distrito y concelho a partir de dichos códigos. En el caso de cpes, se podría valorar su recuperación mediante registros equivalentes o una imputación por grupo si existe suficiente evidencia. Si no existe una forma fiable de recuperar estos valores, podrían mantenerse como NaN o eliminarse únicamente si interfieren con el análisis.

Los valores nulos de freguesia no parecen especialmente relevantes, ya que el proyecto se enfoca principalmente a nivel de concelho. Por tanto, no sería necesario imputarlos salvo que posteriormente se quiera realizar un análisis específico a nivel de freguesia.

Para los valores faltantes de tipo_de_tecnologia y nivel_de_tensao, aunque son muy pocos en relación con el tamaño total del dataset (21 y 15 respectivamente), se podría realizar una imputación por grupo únicamente si la categoría correspondiente es inequívoca. En caso contrario, sería preferible mantenerlos como NaN o eliminar esas filas si dichas variables son imprescindibles para el análisis.

Respecto a los valores extremos, si se comprueba que son reales se mantendrán. Si se determina que son errores, se intentará recuperar el valor correcto a partir de registros equivalentes o de la fuente original. Si esto no es posible, se valorará una imputación por grupo adecuada o, como última opción, la eliminación del registro.

En cuanto a la granularidad, se comprobó que una observación no se define únicamente por año, mes y freguesia. La combinación también depende de codigo_postal, tipo_de_tecnologia, nivel_de_tensao y escalao_de_potencia_instalada. Además, en algunos registros freguesia aparece como NaN mientras codfreguesia sí está informado, por lo que resulta más fiable utilizar el código territorial para identificar la unidad geográfica.

Por tanto, la granularidad original parece venir determinada por año + mes + codfreguesia + codigo_postal + tipo_de_tecnologia + nivel_de_tensao + escalao_de_potencia_instalada. Posteriormente será necesario agregar estos datos al nivel mes + concelho para combinarlos con el resto de datasets.
