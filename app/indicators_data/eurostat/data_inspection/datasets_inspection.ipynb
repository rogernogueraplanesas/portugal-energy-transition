## Observaciones para el dataset 'nrg_ind_ren'

En este primer dataset de Eurostat se muestra el porcentaje sobre el total de energía consumida en bruto que corresponde a fuentes de energía renovable, las propias unidades quedan confirmadas dentro de los propios datos por la columna 'unit'.

Se trata de un dataset corto de apenas 167 filas y 26 columnas, muchas de ellas relativas al año para el cual se muestra la métrica correspondiente.

Se observan distintos valores nulos por columnas excepto para el periodo de 2021 a 2024.
Es posible que el repunte de nulos de 2025 (104 de 167 filas) venga dado por estar aún actualizándose; habrá que decidir si obviar este año una vez se confirme la hipótesis.
Para el resto de años lo más prudente será si decidir entre eliminar aquellos países con menos influencia en el análisis que presenten nulos o si simplemente dejarlos como valores nulos.

Si bien no hay duplicados globales, vemos que la granularidad no es simplemente por país, de modo que el dataset no muestra un único valor porcentual de consumo de energía renovable sobre consumo total.
Se comprueba que para cada país hay un desglose de hasta un máximo de 6 subcategorías en función del balance/sector (nrg_bal).
Para cada país se presenta el valor porcentual de consumo renovable sobre consumo bruto total con el código 'REN', luego se subdivide este consumo en otros porcentajes que lo conforman:

'REN_ELC': Muestra el porcentaje de renovables sobre el consumo total solo de electricidad.
'REN_HEAT_CL': En este caso muestra el porcentaje sobre el total de calefacción y refrigeración.
'REN_TRA': Aplica sobre el consumo total en transporte.
'REN_HEAT_CL_WHC' y 'REN_WHC_DHEAT_DCL' están relacionados con el consumo de frío/calor residual.

Para este proyecto es muy probable que simplemente se use el valor genérico de 'REN' por fecha y país, dicho lo cual, hasta que se cree un segundo dataset que solo tenga en cuenta estos factores, la granularidad del dataset se mantiene en país y balance/sector del consumo, siendo el objetivo que la granularidad acabe marcada por el año (ahora son columnas), país y balance/sector.

Cabe la posibilidad de usar 'REN_ELC' como variable adicional dentro de la hipótesis número 5 dado su buen encaje con el dataset '26-centrais' de E-Redes.

El tipo de datos en este caso debería ser distinto para 'nrg_bal' y 'geo\TIME_PERIOD', las cuales deberían ser categóricas, aunque la segunda podría ser texto. Por otro lado, unit puede ser categórica ya que al ser 'PC' para todos los registros, no aporta variabilidad.

La franja temporal de este dataset abarca desde 2004 a 2025, a la espera de decidir si el corte aplica a finales de 2024 por falta de datos en 2025.

## Observaciones para el dataset 'nrg_pc_204'

Las unidades de las métricas vienen incluídas en la columna 'unit' siendo que los precios se expresan en kWh.

El dataset presenta 2196 filas y 45 columnas, muchas de ellas muestran métricas semestrales en un rango de 18 años.

Se observa que la columna 'nrg_cons' indica la banda de consumo que abarca las siguientes opciones:

KWH_LT1000

KWH1000-2499

KWH2500-4999

KWH5000-14999

KWH_GE15000

TOT_KWH

Es de mucha utilidad tener el desglose por franjas además del total puesto que se analilzará el tipo de consumo y no solo la cantidad.

La columna 'tax' indica el tipo de impuesto incluido en el precio marcado, 'I_TAX' incluye todos los impuestos y tasas, 'X_TAX' no incluye impuestos y 'X_VAT' incluye impuestos y tasas pero no el IVA.

Para este proyecto se filtrará de forma que se tenga únicamente en cuenta el precio con impuestos, dado que el objetivo es determinar el precio que paga realmente el consumidor portugués.

Por otro lado, la columna currency indica el valor en euros y en la moneda nacional (en este caso coincide). Además hay una tercera opción que es el 'PPS' que indica el 'Purchasing Power Standard', una unidad ajustada por poder adquisitivo que es muy útil para hacer comparativas entre países.

En cuanto a tipos, cabría transformar a categóricas las columnas 'freq', 'siec' 'nrg_cons' (en este caso con orden), 'unit', 'tax', 'currency' y 'geo\\TIME_PERIOD'.

'freq', 'siec' y 'unit' se mantienen constantes por lo que no son muy relevantes. En cuanto a 'nrg_cons', habría que tener en cuenta que 'TOT_KWH' es un agregado.


Se observan bastantes nulos para las columnas con métricas y ninguno para las demás. Si bien se puede valorar reducir el rango temporal en función de como avance el análisis, la idea sería mantener los nulos o estudiar si hay alguna forma de imputarlos que sea fiable.

No hay duplicados globales y se determina que las dimensiones que marcan la granularidad son 'geo\\TIME_PERIOD', 'ngr_cons', 'tax' y 'currency'. Si al final se pasara a formato largo la granularidad sería de 'periodo', 'país', 'nrg_cons', 'tax' y 'currency', aunque no será el objetivo inicialmente.
